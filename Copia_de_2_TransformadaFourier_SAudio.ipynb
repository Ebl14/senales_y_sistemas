# Transformada de Fourier

 - Se presentan los conceptos básicos de la transformada de Fourier, en sus versiones continua, tiempo discreto y transformada discreta (extendida a su versión rápida).

- La transformada de Fourier se puede entender como la extensión de la serie de Fourier para $t\in(-\infty,+\infty)$:

\begin{equation}
X(\omega) = \mathscr{F} \{x(t) \} = \int_{-\infty}^{+\infty}{x(t)e^{-j\omega t}dt}.
\end{equation}

- La respectiva transformada inversa (extensión de la reconstrucción mediante serie de Fourier con $t\in(-\infty,+\infty)$) se define como:

\begin{equation}
x(t) = \mathscr{F}^{-1} \{X(\omega) \} = \frac{1}{2\pi}\int_{-\infty}^{+\infty}{X(\omega)e^{j\omega t}d\omega}.
\end{equation}

- Cabe resaltar entonces que el concepto de espectro en frecuencia para la transformada de Fourier asume que $\omega \in (-\infty,+\infty).$

- En resumen, la serie de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento finito $t\in [-T/2,T/2]$ o $t\in [t_i,t_f]$, en espectro en frecuencia discreto $c_n \in \mathbb{R},\mathbb{C}$, con $n\in \mathbb{Z}.$

- La transformada de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento infinito de tiempo $t\in (-\infty,+\infty)$, en espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R},\mathbb{C}$, con $\omega\in (-\infty,+\infty).$


- Las propiedades principales de la transformada de Fourier son útiles para analizar el espectro en fenómenos físicos clásicos (Ver material apoyo [propiedades de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Propiedades_TransformadaFourier.pdf)).

- Además, algunas de las transformadas básicas se pueden encontrar en las [Tablas de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Tablas_TransformadaFourier.pdf).

# Transformada de Fourier en tiempo discreto

- A partir de la transformada de Fourier, se puede analizar también una señal en tiempo discreto $x[k]\in \mathbb{R} ,\mathbb{C},$ con $k\in\mathbb{Z}$, con el fin de calcular su espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R} ,\mathbb{C}$, pero con $\omega$ continua en el intervalo $\omega\in [-\pi,+\pi]$ o $\omega\in [0,2\pi],$ dado que las bases de representación requieren restringir la codificación de la señal $x[k]$ utilizando funciones seno y coseno originales según la expresión:

\begin{equation}
X(\omega) = \sum_{k\in\mathbb{Z}}{x[k]e^{-j\omega k}},
\end{equation}

- **Nota**: Recuerde que cos/sen en tiempo discreto presentan periodicidad de $2\pi$ en la frecuencia angular $\cos[(\omega\pm r 2\pi)k]=\cos[\omega k]$; con $r,k \in \mathbb{Z}$.

- A la expresión anterior se le conoce comunmente como la trasformada de Fouirer en tiempo discreto (Discrete time Fourier transform - DTFT).
y su respectiva transformada inversa (iDTFT) se expresa como:


\begin{equation}
x[k] = \frac{1}{2\pi}\int_{-\pi}^{+\pi}{X(\omega)e^{j\omega k}d\omega}=\frac{1}{2\pi}\int_{0}^{+2\pi}{X(\omega)e^{j\omega k}d\omega}
\end{equation}


# Transformada discreta de Fourier y transformada rápida de Fourier

- Sin embargo, tanto para la serie de Fourier, como para la transformada (continua y discreta en tiempo), se require de la resolución analítica de integrales; además, de la suposición de intervalos infinitos en tiempo y frecuencia (para el caso de la transformada de Fourier), los cuales no son viables ni implementables en la práctica.

- Por consiguiente, el concepto de espectro en frecuencia se extiende a sistemas de proceso de digital (tiempo discreto y  frecuencia discreta), mediante el cálculo de la transformada discreta de Fourier (Discrete Fourier Transform - DFT).

- En este caso, la transformada discreta se define como:

\begin{equation}
X[n] = \sum_{k=0}^{K-1}{x[k]e^{-jk\frac{2\pi n}{K}}},
\end{equation}

**Nota**: observe que el ángulo en la función euler $\theta_n = \frac{k 2\pi n}{K}$se restringe al intervalo $[0,2\pi)$ para $k\in\{0,1,\dots,K-1\}$, dado que se trabaja con funciones cos/sen en tiempo discreto.

- La DFT es una discretización en frecuencia de la DTFT:

\begin{equation}
X[n] = X(\omega)\Big|_{\omega=\frac{2\pi n}{K}},
\end{equation}

- Respecto a la serie de Fourier, la DFT se puede relacionar mediante:

\begin{equation}
X[n] = K c_n,
\end{equation}
asumiendo señal en tiempo discreto y $\omega_o = \frac{2\pi}{K}$ en representación por serie.

- Por otra parte, la transforma discreta de Fourier inversa (iDFT) se define como:

\begin{equation}
x[k] = \sum_{n=0}^{K-1}{X[n]e^{jk\frac{2\pi n}{K}}}.
\end{equation}

Nuevamente, note que las variables tiempo y frecuencia discretas $n,k\in\{0,1,\dots,K-1\}.$ Dado que la frecuencia fundamental se restringe como $\omega_o=\frac{2\pi}{K},$ las bases de representación en tiempo discreto codificaran frecuencias angulares en el rango $[0,2\pi)$ (cosenos y senos originales).

- No obstante, en la práctica el cálculo de las sumatorias es costoso computacionalmente, a medida que $K$ aumenta. Por consiguiente, se cuentan con paquetes computacionales que optimizan el cálculo de la DFT de forma rápida, lo que conlleva a la comunmente conocida transformada rápida de Fourier ([Fast Fourier Transform - FFT](https://es.wikipedia.org/wiki/Transformada_r%C3%A1pida_de_Fourier)).

# Ejemplo:

- Se presenta el cálculo del espectro mediante DFT (`fft` de numpy) para la señal:


\begin{equation}
x(t) = \begin{cases} A, & -T/2 \leq t < 0 \\ -A, & 0 \leq t < T/2 \end{cases}
\end{equation}

````python
# Para graficar
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")

mpl.rc('axes', labelsize=14)
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)

A = 1 # amplitud de la señal
T = 5 # segundos de gráfica
Fs = 1000 # frecuencia de muestreo definida
tv = np.arange(-T/2,T/2,1/Fs) #vector de tiempo generado
Nm = len(tv) #numero de muestras requeridas

#se crea la señal A y -A concatenando
xo = np.concatenate((A*np.ones(int(Nm/2)), -A*np.ones(int(Nm/2))), axis=0)
#graficar señal original
plt.plot(tv,xo,'r',linewidth=4)
plt.grid()
plt.xlabel("t[s]",fontsize = 14)
plt.ylabel("x(t)",fontsize = 14)
plt.show()
````

````python
#cálculo de la FFT de la señal signo invertido
vf = np.fft.fftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.fft(xo) #cálculo FFT

plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/len(vf),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.show()
````

**Nota**: La función `np.fft.fftfreq` cálcula el equivalente en Hz de cada armónico, teniendo en cuenta que en tiempo discreto:

$$\theta_n = \frac{k 2\pi n}{K} = k \omega_n = k n \omega_o$$

Además, en el proceso de discretización ($F_s$ frecuencia de muestreo):

$$\omega_n t = 2\pi F_n t =  \frac{2\pi F_n k}{F_s} = 2 \pi f n k = \frac{2 \pi n k}{K}$$

donde $f=\frac{1}{K}$ es equivalente a la frecuencia en Hz en tiempo discreto.

Así:

$$\frac{2\pi F_n k}{F_s}=\frac{2 \pi n k}{K}$$

y:

$$F_n = \frac{F_s n}{K} [Hz]$$




- Note que:

$x(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega$

$x^*(t) = \left( \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega\right)^*=\frac{1}{2\pi}\int_{-\infty}^\infty X^*(\omega) e^{-j\omega t} d\omega$

Haciendo $\omega' = - \omega$:


$x^*(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X^*(-\omega') e^{j\omega' t} d\omega'$

Entonces:

$\mathscr{F}\{x^*(t)\} = X^*(-\omega) = \mathscr{Re}\{X(-\omega) \} -j\mathscr{Im}\{X(-\omega) \} $

Y

$\mathscr{F}\{x(t)\} = X(\omega) = \mathscr{Re}\{X(\omega) \} +j\mathscr{Im}\{X(\omega) \} $

Por lo tanto:

$\mathscr{Re}\{X(\omega) \}=\mathscr{Re}\{X(-\omega) \}$
$\mathscr{Im}\{X(\omega) \}=-\mathscr{Im}\{X(-\omega) \}$

-Dado que el espectro es simétrico para señales reales ($x(t)\in \mathbb{R}$), se sugiere utilizar la expresión simplificada `rfft`:

````python
vf = np.fft.rfftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.rfft(xo) #cálculo FFT
plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/(2*len(vf)),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.grid()
plt.show()
````

- La reconstrucción de x[k] se puede obtener mediante la transformada inversa:

````python
xe = np.fft.irfft(Xw) #transformada inversa
plt.plot(tv,xe)
plt.title(r'Reconstrucción completa')
plt.xlabel(r't',fontsize = 14)
plt.ylabel(r'$x[n]$',fontsize = 14)
plt.show()
````

- Se pueden escoger o ponderar los coeficientes para filtrar (alterar) la reconstrucción de la señal.

````python
Xwf = Xw.copy()
Xwf[10:] = 0 #solo se tienen en cuenta los 10 primeros armónicos
xe2 = np.fft.irfft(Xwf)
plt.plot(tv,xe2)
plt.show()
````

#  Modelo de Cálculo de Armónicos Usando FFT

La **Transformada Rápida de Fourier (FFT)** es un algoritmo eficiente para calcular la **Transformada Discreta de Fourier (DFT)** de una señal. Se utiliza ampliamente para analizar el contenido en frecuencia de señales periódicas o cuasiperiódicas, permitiendo identificar y cuantificar sus **componentes armónicos**.

##  Representación Matemática

La **DFT** de una señal discreta \( x[n] \) de longitud \( N \) está dada por:

$$[
X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j\frac{2\pi}{N}kn}, \quad k = 0, 1, 2, ..., N-1
]$$

Donde:

- \( X[k] \): Coeficiente complejo del armónico \( k \)-ésimo.  
- \( N \): Número total de muestras.  
- \( x[n] \): Valor de la señal en el tiempo discreto \( n \).  
- \( k \): Índice de frecuencia correspondiente al \( k \)-ésimo armónico.

La **FFT** optimiza este cálculo reduciendo la complejidad computacional de \( O(N^2) \) a \( O(N \log N) \).

##  Modelo para el Cálculo de Armónicos

1. **Adquisición de la señal**:  
   Se toma una señal periódica o cuasiperiódica \( x(t) \), y se discretiza a través de muestreo con frecuencia \( f_s \), obteniendo \( x[n] \).

2. **Aplicación de ventana** *(opcional)*:  
   Para reducir el efecto de fugas espectrales, se aplica una ventana como Hamming o Hann antes de calcular la FFT.

3. **Cálculo de la FFT**:

```python
import numpy as np

X = np.fft.fft(x)                # Transformada rápida de Fourier
freqs = np.fft.fftfreq(len(x), d=1/fs)  # Vector de frecuencias
## 🔍 Extracción de Armónicos
```
- La **magnitud** de los armónicos está dada por:

$$[
|X[k]|
]$$

- La **fase** de los armónicos está dada por:

$$[
\arg(X[k])
]$$

- Cada índice \( k \) se asocia con la frecuencia:

$$[
f_k = \frac{k \cdot f_s}{N}
]$$

Donde:

- \( f_k\ ): Frecuencia del armónico \( k \)-ésimo
- \( f_s \): Frecuencia de muestreo  
- \( N \): Número total de muestras  


````python
import numpy as np
import time
import matplotlib.pyplot as plt

def dft_manual(x):
    """Calcula la DFT usando la sumatoria directa."""
    N = len(x)
    X = np.zeros(N, dtype=complex)
    for k in range(N):
        for n in range(N):
            X[k] += x[n] * np.exp(-2j * np.pi * k * n / N)
    return X

signal_lengths = [4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096]
manual_dft_times = []
fft_times = []

for N in signal_lengths:
    # Crear una señal de prueba
    x = np.random.rand(N)

    # Medir tiempo para la DFT manual
    start_time = time.time()
    dft_manual(x)
    end_time = time.time()
    manual_dft_times.append(end_time - start_time)

    # Medir tiempo para la FFT de numpy
    start_time = time.time()
    np.fft.fft(x)
    end_time = time.time()
    fft_times.append(end_time - start_time)

plt.figure(figsize=(10, 6))
plt.plot(signal_lengths, manual_dft_times, marker='o', label='DFT Manual')
plt.plot(signal_lengths, fft_times, marker='o', label='FFT (numpy)')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Longitud de la señal')
plt.ylabel('Tiempo de ejecución (segundos)')
plt.title('Comparación de tiempo de cómputo: DFT vs FFT')
plt.legend()
plt.grid(True)
plt.show()
````

## Frecuencia de -3 dB y su relación con las frecuencias de corte en el filtrado espectral

La **frecuencia de -3 dB** es aquella en la cual la **potencia** de una señal se reduce a la mitad respecto de su valor máximo en la banda pasante de un sistema. En términos de amplitud, esto corresponde a una reducción al 70.7% del valor máximo.

Dado que la ganancia en decibelios se calcula mediante:

$$
\text{Ganancia (dB)} = 20 \cdot \log_{10}\left(\frac{V_{\text{salida}}}{V_{\text{entrada}}}\right)
$$

Una atenuación de -3 dB implica:

$$
\frac{V_{\text{salida}}}{V_{\text{entrada}}} = 10^{-3/20} \approx 0.707
$$

Como la potencia es proporcional al cuadrado de la amplitud:

$$
\left(0.707\right)^2 \approx 0.5
$$

Esto significa que a -3 dB, la **potencia se reduce al 50%**.

---

### Relación con las frecuencias de corte

En el análisis de filtros (ya sean analógicos o digitales), las **frecuencias de corte** se definen como aquellas en las que la ganancia cae a -3 dB respecto al valor máximo. Estas frecuencias marcan la transición entre bandas:

- **Banda pasante**: donde las frecuencias se transmiten con poca o ninguna atenuación.
- **Banda de atenuación o rechazo**: donde las frecuencias se reducen considerablemente.

Dependiendo del tipo de filtro:

- **Filtro pasa bajos**: frecuencia de corte marca el inicio de la atenuación para frecuencias más altas.
- **Filtro pasa altos**: frecuencia de corte marca el inicio de la atenuación para frecuencias más bajas.
- **Filtro pasa banda**: tiene dos frecuencias de corte (-3 dB), delimitando el rango que se deja pasar.
- **Filtro elimina banda (notch)**: también tiene dos frecuencias de corte que delimitan la banda que se atenúa.

---

### Conclusión

La frecuencia de -3 dB es un estándar práctico y teórico para definir los límites efectivos de paso de un filtro. También permite caracterizar la **respuesta en frecuencia** de un sistema, lo cual es fundamental en el diseño y análisis de circuitos y algoritmos de procesamiento de señales.


````python
# Para instalar yt-dlp y ffmpeg si aún no están
# !python3 -m pip install --force-reinstall https://github.com/yt-dlp/yt-dlp/archive/master.tar.gz
# !apt update && apt install ffmpeg

# Debes incluir el link del video a procesar:
link="https://www.youtube.com/watch?v=qWCup_EZSWE" # Cambia este link por el de tu canción favorita
!yt-dlp --extract-audio -o "audio" --audio-format mp3 {link}

# Se convierte el audio a .wav para procesar:
!ffmpeg -y -i audio.mp3 output.wav

# Librería para manipulación de archivos de audio
# !pip install soundfile

import soundfile as sf
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio

#lee archivos wav
nombre_out = "output.wav"
x, fs = sf.read(nombre_out)
print('Frecuencia de muestreo %.2f[Hz]\naudio %s' % (fs,nombre_out))
print('Forma del audio:', x.shape)

# Seleccionar 5 segundos del audio (o ajustar según la duración disponible)
ns = 5 # Duración en segundos a procesar
start_sample = 0
end_sample = int(fs * ns)
xs = x[start_sample:end_sample, :]

print('Forma del segmento de audio a procesar:', xs.shape)

# Asegurarse de que el segmento no exceda la longitud total del audio
if xs.shape[0] == 0 and x.shape[0] > 0:
    print("Error: El segmento solicitado excede la duración del audio. Procesando la duración completa.")
    xs = x
    end_sample = x.shape[0]
    ns = x.shape[0] / fs
elif xs.shape[0] == 0 and x.shape[0] == 0:
     print("Error: No se pudo cargar el audio.")
     # Puedes salir o manejar el error de otra manera
     exit()

# Cálculo de la FFT del segmento de audio
Xw = np.fft.rfft(xs, axis=0)
vf = np.fft.rfftfreq(xs.shape[0], 1/fs)

# --- Implementación de los filtros ---

# Frecuencias de corte (ajusta estos valores según tu audio y el efecto deseado)
# Estas son frecuencias de ejemplo en Hz
fc_pb = 500      # Frecuencia de corte para filtro pasa-bajas
fc_pa = 2000     # Frecuencia de corte para filtro pasa-altas
f1_pb = 300      # Frecuencia de inicio de banda para pasa-banda
f2_pb = 1500     # Frecuencia de fin de banda para pasa-banda
f1_rb = 800      # Frecuencia de inicio de banda para rechaza-banda
f2_rb = 1200     # Frecuencia de fin de banda para rechaza-banda


print("\nAplicando filtros:")

# 1. Filtro Pasa-Bajas
print("- Aplicando filtro Pasa-Bajas (corte en", fc_pb, "Hz)")
Xw_pb = Xw.copy()
# Creamos una máscara booleana: True para las frecuencias a atenuar, False para pasar
mask_pb = vf > fc_pb
Xw_pb[mask_pb, :] = 0 # Poner a cero las frecuencias por encima de fc_pb
xe_pb = np.fft.irfft(Xw_pb, axis=0)

plt.figure(figsize=(10, 4))
plt.plot(vf, abs(Xw_pb))
plt.title(f'Espectro Audio Filtrado Pasa-Bajas (corte {fc_pb} Hz)')
plt.xlabel(r'$f[Hz]$', fontsize = 14)
plt.ylabel(r'$|X[n]|$', fontsize = 14)
plt.legend(('canal 1','canal 2'))
plt.grid()
plt.show()
print("Audio Filtrado Pasa-Bajas:")
display(Audio(xe_pb.T, rate=fs))


# 2. Filtro Pasa-Altas
print("- Aplicando filtro Pasa-Altas (corte en", fc_pa, "Hz)")
Xw_pa = Xw.copy()
# Creamos una máscara booleana: True para las frecuencias a atenuar (por debajo de fc_pa)
mask_pa = vf < fc_pa
Xw_pa[mask_pa, :] = 0 # Poner a cero las frecuencias por debajo de fc_pa
xe_pa = np.fft.irfft(Xw_pa, axis=0)

plt.figure(figsize=(10, 4))
plt.plot(vf, abs(Xw_pa))
plt.title(f'Espectro Audio Filtrado Pasa-Altas (corte {fc_pa} Hz)')
plt.xlabel(r'$f[Hz]$', fontsize = 14)
plt.ylabel(r'$|X[n]|$', fontsize = 14)
plt.legend(('canal 1','canal 2'))
plt.grid()
plt.show()
print("Audio Filtrado Pasa-Altas:")
display(Audio(xe_pa.T, rate=fs))


# 3. Filtro Pasa-Bandas
print(f"- Aplicando filtro Pasa-Bandas ({f1_pb}-{f2_pb} Hz)")
Xw_pb_band = Xw.copy()
# Creamos una máscara booleana: True para las frecuencias fuera de la banda ( < f1_pb O > f2_pb)
mask_pb_band = (vf < f1_pb) | (vf > f2_pb)
Xw_pb_band[mask_pb_band, :] = 0 # Poner a cero las frecuencias fuera de la banda
xe_pb_band = np.fft.irfft(Xw_pb_band, axis=0)

plt.figure(figsize=(10, 4))
plt.plot(vf, abs(Xw_pb_band))
plt.title(f'Espectro Audio Filtrado Pasa-Bandas ({f1_pb}-{f2_pb} Hz)')
plt.xlabel(r'$f[Hz]$', fontsize = 14)
plt.ylabel(r'$|X[n]|$', fontsize = 14)
plt.legend(('canal 1','canal 2'))
plt.grid()
plt.show()
print("Audio Filtrado Pasa-Bandas:")
display(Audio(xe_pb_band.T, rate=fs))


# 4. Filtro Rechaza-Bandas
print(f"- Aplicando filtro Rechaza-Bandas ({f1_rb}-{f2_rb} Hz)")
Xw_rb_band = Xw.copy()
# Creamos una máscara booleana: True para las frecuencias dentro de la banda a rechazar ( >= f1_rb Y <= f2_rb)
mask_rb_band = (vf >= f1_rb) & (vf <= f2_rb)
Xw_rb_band[mask_rb_band, :] = 0 # Poner a cero las frecuencias dentro de la banda a rechazar
xe_rb_band = np.fft.irfft(Xw_rb_band, axis=0)

plt.figure(figsize=(10, 4))
plt.plot(vf, abs(Xw_rb_band))
plt.title(f'Espectro Audio Filtrado Rechaza-Bandas ({f1_rb}-{f2_rb} Hz)')
plt.xlabel(r'$f[Hz]$', fontsize = 14)
plt.ylabel(r'$|X[n]|$', fontsize = 14)
plt.legend(('canal 1','canal 2'))
plt.grid()
plt.show()
print("Audio Filtrado Rechaza-Bandas:")
display(Audio(xe_rb_band.T, rate=fs))

print("\nProceso de filtrado completado.")
````
