**Series de Fourier**

*  Las Series de Fourier se utilizan para representar funciones periódicas como una suma de senos y cosenos (o exponenciales complejas).

**Serie de Fourier Exponencial:**

*  Representa una función periódica $f(t)$ con período $T_0$ como: $f(t) = \sum_{n=-\infty}^{\infty} c_n e^{j n \omega_0 t}$ donde $\omega_0 = 2\pi / T_0$ y los coeficientes $c_n$ se calculan como: $c_n = \frac{1}{T_0} \int_{t_0}^{t_0+T_0} f(t) e^{-j n \omega_0 t} dt$


**Serie de Fourier Trigonométrica:**

*  Representa una función periódica $f(t)$ con período $T_0$ como: $f(t) = a_0 + \sum_{n=1}^{\infty} (a_n \cos(n \omega_0 t) + b_n \sin(n \omega_0 t))$ donde $\omega_0 = 2\pi / T_0$ y los coeficientes se calculan como: $a_0 = \frac{1}{T_0} \int_{t_0}^{t_0+T_0} f(t) dt$ $a_n = \frac{2}{T_0} \int_{t_0}^{t_0+T_0} f(t) \cos(n \omega_0 t) dt$ $b_n = \frac{2}{T_0} \int_{t_0}^{t_0+T_0} f(t) \sin(n \omega_0 t) dt$


**Serie de Fourier Compacta:**
*  Es otra forma de la serie trigonométrica, expresando la suma de senos y cosenos como una sola sinusoide con fase.


**Transformada de Fourier (Continuous-Time Fourier Transform - CTFT)**

*   La Transformada de Fourier se utiliza para representar funciones no periódicas (o aperiódicas) como una integral de exponenciales complejas. Representa la función en el dominio de la frecuencia.

**Definición:**

  *   La transformada de Fourier de una función $f(t)$ es: $F(\omega) = \int_{-\infty}^{\infty} f(t) e^{-j \omega t} dt$


**Transformada Inversa de Fourier:**

*   La transformada inversa es: $f(t) = \frac{1}{2\pi} \int_{-\infty}^{\infty} F(\omega) e^{j \omega t} d\omega$

**Transformada de Fourier en Tiempo Discreto (Discrete-Time Fourier Transform - DTFT)**

* La DTFT se utiliza para representar secuencias discretas (señales definidas solo en instantes de tiempo discretos) como una integral de exponenciales complejas en el dominio de la frecuencia. El dominio de la frecuencia es continuo y periódico.


**Definición:**
* La DTFT de una secuencia $x[n]$ es: $X(e^{j\omega}) = \sum_{n=-\infty}^{\infty} x[n] e^{-j \omega n}$

**Transformada Inversa de Fourier en Tiempo Discreto:**
* La transformada inversa es: $x[n] = \frac{1}{2\pi} \int_{-\pi}^{\pi} X(e^{j\omega}) e^{j \omega n} d\omega$

**Transformada Discreta de Fourier (Discrete Fourier Transform - DFT)**

* La DFT se utiliza para representar secuencias discretas y finitas como una suma finita de exponenciales complejas discretas. Se utiliza comúnmente en procesamiento digital de señales y se puede calcular eficientemente usando el algoritmo de la Transformada Rápida de Fourier (FFT).

**Definición:**
* La DFT de una secuencia de $N$ muestras $x[n]$ es: $X[k] = \sum_{n=0}^{N-1} x[n] e^{-j \frac{2\pi}{N} k n}$, para $k = 0, 1, \ldots, N-1$

**Transformada Inversa Discreta de Fourier:**
* La transformada inversa es: $x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j \frac{2\pi}{N} k n}$, para $n = 0, 1, \ldots, N-1$

**Semejanzas:**

* Todas representan señales en términos de sus componentes de frecuencia.
* Todas utilizan exponenciales complejas (o senos y cosenos) como base para la representación.
* Conceptualmente, buscan descomponer una señal en sus "frecuencias constituyentes".

**Diferencias:**

* **Dominio de la Señal:**
    * Las Series de Fourier trabajan con señales continuas y periódicas. La CTFT trabaja con señales continuas y no periódicas. La DTFT trabaja con señales discretas y no periódicas. La DFT trabaja con señales discretas y finitas.
* **Dominio de la Frecuencia:**
    * Para las Series de Fourier, el espectro de frecuencia es discreto. Para la CTFT, el espectro de frecuencia es continuo y no periódico. Para la DTFT, el espectro de frecuencia es continuo y periódico. Para la DFT, el espectro de frecuencia es discreto y finito.
* **Tipo de Representación:**
    * Las Series de Fourier representan la señal como una suma infinita. La CTFT y la DTFT representan la señal como una integral. La DFT representa la señal como una suma finita.
* **Aplicación:**
    * Las Series de Fourier se usan para analizar señales periódicas. La CTFT se usa para analizar señales continuas no periódicas. La DTFT se usa para analizar señales discretas (pero el espectro es continuo). La DFT se usa para analizar digitalmente señales discretas y finitas, y es la base para la implementación práctica de análisis de frecuencia en sistemas digitales.

# Combinación de Tiempo y Espectro en Señales

Cuando se combinan las características del **tiempo** (continuo o discreto) con las del **espectro** (continuo o discreto), se obtienen **cuatro combinaciones posibles**. Estas combinaciones describen cómo una señal puede comportarse tanto en el dominio del tiempo como en el dominio de la frecuencia. A continuación se detallan:

---

###  1. Señal en **tiempo continuo** con **espectro continuo**

- **Ejemplo**: Señales analógicas como una sinusoide pura o una suma de senos.
- **Transformada usada**: Transformada de Fourier continua.
- **Aplicación típica**: Audio analógico, señales de radio.
- **Características**: Energía distribuida continuamente en frecuencia.

---

###  2. Señal en **tiempo continuo** con **espectro discreto**

- **Ejemplo**: Señales periódicas en tiempo continuo (como una sinusoide de frecuencia fija).
- **Transformada usada**: Serie de Fourier.
- **Aplicación típica**: Osciladores, sistemas con señales periódicas.
- **Características**: Componentes armónicos en frecuencias discretas.

---

###  3. Señal en **tiempo discreto** con **espectro continuo**

- **Ejemplo**: Señales no periódicas en tiempo discreto (como una señal muestreada que no es periódica).
- **Transformada usada**: Transformada de Fourier discreta en tiempo (DTFT).
- **Aplicación típica**: Señales digitales capturadas de fenómenos naturales.
- **Características**: El espectro es continuo dentro del intervalo $[-\pi, \pi]$.

---

###  4. Señal en **tiempo discreto** con **espectro discreto**

- **Ejemplo**: Señales periódicas en tiempo discreto.
- **Transformada usada**: Serie de Fourier discreta (DFS) o DFT.
- **Aplicación típica**: Procesamiento digital de señales periódicas, compresión.
- **Características**: Espectro con un número finito de componentes.

# Fast Fourier Transform (FFT) y su utilidad

La **Fast Fourier Transform (FFT)** o **Transformada Rápida de Fourier** es un **algoritmo eficiente** para calcular la **Transformada Discreta de Fourier (DFT)** de una secuencia. Mientras que la DFT permite transformar una señal del dominio del tiempo al dominio de la frecuencia, su cálculo directo es **computacionalmente costoso** para señales largas.

---

##  ¿Qué es la DFT?

La **Transformada Discreta de Fourier** para una secuencia de $N$ puntos se define como:

$$[
X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j \frac{2\pi}{N}kn}, \quad k = 0, 1, ..., N-1
]$$

Esto requiere realizar $N$ sumas, cada una con $N$ multiplicaciones ⇒ **costo computacional: $\mathcal{O}(N^2)$**.

---

# ¿Qué hace la FFT?

La **FFT reduce drásticamente el número de operaciones necesarias** para obtener la DFT, pasando de $\mathcal{O}(N^2)$ a **$\mathcal{O}(N \log N)$**.

Para ello, aprovecha propiedades de simetría y periodicidad de los factores exponenciales (llamados *twiddle factors*) y divide recursivamente el problema en subproblemas más pequeños.

---

# ¿Cómo funciona la FFT? (Algoritmo de Cooley-Tukey)

El algoritmo **más común** de FFT es el **de Cooley-Tukey**, que funciona mejor cuando $N$ es una potencia de 2. La idea principal es:

1. **Dividir** la secuencia original $x[n]$ en dos subsecuencias:
   - Los elementos **pares**: $x[0], x[2], x[4], \dots$
   - Los elementos **impares**: $x[1], x[3], x[5], \dots$

2. Aplicar recursivamente la FFT a ambas subsecuencias (de longitud $N/2$).

3. Combinar los resultados usando los factores de rotación:

$$[
X[k] = E[k] + W_N^k \cdot O[k] \\
X[k + N/2] = E[k] - W_N^k \cdot O[k]
]$$

donde:
- $E[k]$ es la FFT de la parte par.
- $O[k]$ es la FFT de la parte impar.
- $W_N^k = e^{-j\frac{2\pi k}{N}}$ es el *twiddle factor*.

---

# Complejidad Computacional

| Método        | Complejidad               |
|---------------|---------------------------|
| DFT directa   | $\mathcal{O}(N^2)$        |
| FFT (Cooley-Tukey) | $\mathcal{O}(N \log N)$ |

Esto significa que para señales largas (por ejemplo, $N = 1024$, $2048$, etc.), la FFT es **muchísimo más rápida** que el cálculo directo de la DFT.

---

# Utilidad de la FFT

- **Procesamiento de señales** en tiempo real (audio, radar, comunicaciones).
- **Análisis espectral** rápido y eficiente.
- **Compresión** y **filtrado** de señales digitales.
- **Reconstrucción de señales** en aplicaciones médicas e industriales.

# Transformada de Fourier de señales específicas

Usamos la **convención de Fourier bilateral**:

$$
\boxed{
X(f) = \int_{-\infty}^{\infty} x(t) e^{-j 2\pi f t} dt
}
$$

---

**a)** $x(t) = e^{-a|t|}, \quad a \in \mathbb{R}^+$

Esta función es par. La TF se calcula como:

$$
X(f) = 2 \int_0^\infty e^{-a t} \cos(2\pi f t) dt = \frac{2a}{a^2 + (2\pi f)^2}
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{e^{-a|t|}\} = \frac{2a}{a^2 + (2\pi f)^2}
}
$$

---

**b)** $x(t) = \cos(2\pi f_c t), \quad f_c \in \mathbb{R}$

Usando la identidad de Euler:

$$
\cos(2\pi f_c t) = \frac{1}{2} \left( e^{j 2\pi f_c t} + e^{-j 2\pi f_c t} \right)
$$

$$
X(f) = \frac{1}{2}\left[\delta(f - f_c) + \delta(f + f_c)\right]
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{\cos(2\pi f_c t)\} = \frac{1}{2}\left[\delta(f - f_c) + \delta(f + f_c)\right]
}
$$

---

**c)** $x(t) = \sin(2\pi f_s t), \quad f_s \in \mathbb{R}$

$$
\sin(2\pi f_s t) = \frac{1}{2j} \left( e^{j 2\pi f_s t} - e^{-j 2\pi f_s t} \right)
$$

$$
X(f) = \frac{1}{2j}\left[\delta(f - f_s) - \delta(f + f_s)\right]
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{\sin(2\pi f_s t)\} = \frac{1}{2j}\left[\delta(f - f_s) - \delta(f + f_s)\right]
}
$$

---

**d)** $x(t) = f(t)\cos(2\pi f_c t), \quad f(t) \in \mathbb{R} \cup \mathbb{C}$

Esta señal **no puede transformarse directamente** sin conocer la forma explícita de $f(t)$. No se puede resolver la integral sin aplicar propiedades como convolución o modulación.

**Resultado: No evaluable directamente sin conocer $f(t)$.**

---

**e)** $x(t) = e^{-a t^2}, \quad a \in \mathbb{R}^+$

Esta es una **gaussiana**. La TF también es gaussiana:

$$
X(f) = \sqrt{\frac{\pi}{a}} \cdot e^{-\frac{(\pi f)^2}{a}}
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{e^{-a t^2}\} = \sqrt{\frac{\pi}{a}} \cdot e^{-\frac{(\pi f)^2}{a}}
}
$$

---

**f)** $x(t) = A \cdot \text{rect}\left(\frac{t}{d}\right), \quad A, d \in \mathbb{R}$

La función $\text{rect}(t/d)$ vale 1 entre $-d/2$ y $d/2$:

$$
X(f) = A \int_{-d/2}^{d/2} e^{-j 2\pi f t} dt = A \cdot d \cdot \text{sinc}(f d)
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\left\{A \cdot \text{rect}\left(\frac{t}{d}\right)\right\} = A \cdot d \cdot \text{sinc}(f d)
}
$$

donde $\text{sinc}(x) = \frac{\sin(\pi x)}{\pi x}$.

a) $\mathcal{F}\left\{ e^{-j\omega_1 t} \cos(\omega_c t) \right\}, \omega_1, \omega_c \in \mathbb{R}$

**Solución:**
Utilizamos la identidad de Euler para el coseno y la propiedad de desplazamiento en frecuencia:

$$
\cos(\omega_c t) = \frac{1}{2}\left(e^{j\omega_c t} + e^{-j\omega_c t}\right)
$$

Luego:

$$
e^{-j\omega_1 t} \cos(\omega_c t) = \frac{1}{2}\left(e^{j(\omega_c - \omega_1)t} + e^{-j(\omega_c + \omega_1)t}\right)
$$

Aplicando la transformada de Fourier:

$$
\mathcal{F}\left\{ e^{-j\omega_1 t} \cos(\omega_c t) \right\} = \frac{1}{2}\left[2\pi\delta(\omega - (\omega_c - \omega_1)) + 2\pi\delta(\omega + (\omega_c + \omega_1))\right]
$$

**Resultado final:**

$$
\pi \delta(\omega - \omega_c + \omega_1) + \pi \delta(\omega + \omega_c + \omega_1)
$$

---

b) $\mathcal{F}\left\{ u(t) \cos^2(\omega_c t) \right\}, \quad \omega_c \in \mathbb{R}$

**Solución:**
Primero simplificamos la expresión usando identidades trigonométricas:

$$
\cos^2(\omega_c t) = \frac{1}{2}\left[1 + \cos(2\omega_c t)\right]
$$

Entonces:

$$
u(t)\cos^2(\omega_c t) = \frac{1}{2}u(t) + \frac{1}{2}u(t)\cos(2\omega_c t)
$$

Aplicando la transformada de Fourier a cada término:

1. $\mathcal{F}\left\{\frac{1}{2}u(t)\right\} = \frac{1}{2}\left[\pi\delta(\omega) + \frac{1}{j\omega}\right]$

2. Para el segundo término usamos la propiedad de modulación:

$$
\mathcal{F}\left\{u(t)\cos(2\omega_c t)\right\} = \frac{1}{2}\left[\mathcal{F}\{u(t)\}(\omega - 2\omega_c) + \mathcal{F}\{u(t)\}(\omega + 2\omega_c)\right]
$$

**Resultado final:**

$$
\frac{\pi}{2}\delta(\omega) + \frac{1}{2j\omega} + \frac{\pi}{4}\left[\delta(\omega - 2\omega_c) + \delta(\omega + 2\omega_c)\right] + \frac{1}{4j(\omega - 2\omega_c)} + \frac{1}{4j(\omega + 2\omega_c)}
$$

---

c) $\mathcal{F}^{-1}\left\{ \frac{7}{\omega^2 + 6\omega + 45} \cdot \frac{10}{(8 + j\omega/3)^2} \right\}$

**Solución:**
Primero completamos cuadrados en el primer denominador:

$$
\omega^2 + 6\omega + 45 = (\omega + 3)^2 + 36
$$

El segundo término ya está factorizado. Podemos expresar la transformada inversa como una convolución:

1. $\mathcal{F}^{-1}\left\{\frac{7}{(\omega+3)^2 + 36}\right\} = \frac{7}{12}e^{-3t}\sin(6t)u(t)$

2. $\mathcal{F}^{-1}\left\{\frac{10}{(8 + j\omega/3)^2}\right\} = 30t e^{-8t}u(t)$ (usando escalamiento en frecuencia)

**Resultado final:**

La transformada inversa es la convolución de estas dos funciones:

$$
f(t) = \frac{35}{4} \int_0^t e^{-3\tau}\sin(6\tau) (t-\tau)e^{-8(t-\tau)} d\tau \quad \text{para } t \geq 0
$$

---

d) $\mathcal{F}\left\{ 3t^3 \right\}$

**Solución:**
Usamos la propiedad de la derivada en frecuencia:

$$
\mathcal{F}\{t^n f(t)\} = j^n \frac{d^n}{d\omega^n} F(\omega)
$$

Para $f(t) = 1$, $F(\omega) = 2\pi\delta(\omega)$. Entonces:

$$
\mathcal{F}\{t^3\} = j^3 \frac{d^3}{d\omega^3} [2\pi\delta(\omega)] = -2\pi j \delta'''(\omega)
$$

**Resultado final:**

$$
6\pi j \delta'''(\omega)
$$

Reconocemos que esta es una función periódica en el dominio de la frecuencia con período $\omega_0 = 2\pi/T$. Podemos expresarla como:

# Modulación por Amplitud con Detección Coherente (DSB-SC)

La **modulación por amplitud con portadora suprimida** (DSB-SC, *Double Sideband Suppressed Carrier*) es un esquema de modulación donde la señal mensaje modula directamente una portadora mediante multiplicación, **eliminando la componente de la portadora** en la señal transmitida. A diferencia de la AM convencional, **requiere detección coherente** en el receptor, es decir, la portadora local debe estar sincronizada en fase y frecuencia para una demodulación correcta.

---

## **Características Clave**

**Señal modulada:**

\[
s(t) = m(t) \cdot \cos(2\pi f_c t)
\]

donde \( m(t) \) es la señal mensaje y \( f_c \) es la frecuencia de la portadora.

**Demodulación:**
1. Multiplicar por \( \cos(2\pi f_c t) \) (portadora local).
2. Aplicar un filtro pasa bajos para recuperar \( m(t) \).

---

### **Ventajas**
- Mayor eficiencia espectral (no se desperdicia potencia en la portadora).
- Ideal para sistemas con restricciones de ancho de banda.

### **Desventajas**
- Requiere sincronización precisa en el receptor (complejidad aumentada).

---

## Aplicaciones
- Comunicaciones de banda lateral única (SSB).
- Radio definida por software (SDR).
- Transmisión de señales analógicas y digitales en entornos con interferencia controlada.

---

## Simulación en Python

El siguiente código genera y visualiza una señal DSB-SC, tanto en el dominio del tiempo como en el de la frecuencia, usando dos tipos de señales mensaje:

- **Pulso rectangular** (para ver el efecto en señales no periódicas).
- **Señal cosenoidal** (para analizar el espectro típico).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import rfft, rfftfreq

# Configuración
fs = 10000       # Frecuencia de muestreo (Hz)
t = np.arange(0, 0.01, 1/fs)  # Vector de tiempo (10 ms)
fc = 1000        # Frecuencia de la portadora (Hz)
mod_index = 1.0  # Índice de modulación (puede ajustarse)

# Señales mensaje
rect_pulse = np.where((t >= 0.003) & (t <= 0.007), 1.0, 0.0)  # Pulso de 4 ms
cos_msg = np.cos(2 * np.pi * 100 * t)  # Señal cosenoidal de 100 Hz

# Modulación DSB-SC
def modulate_dsb_sc(msg):
    carrier = np.cos(2 * np.pi * fc * t)
    return mod_index * msg * carrier

mod_rect = modulate_dsb_sc(rect_pulse)
mod_cos = modulate_dsb_sc(cos_msg)

# Visualización en tiempo y frecuencia
def plot_signal(signal, title):
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6))

    # Dominio del tiempo
    ax1.plot(t, signal, color='b')
    ax1.set_title(f'{title} - Dominio del Tiempo')
    ax1.set_xlabel('Tiempo (s)')
    ax1.set_ylabel('Amplitud')
    ax1.grid(True)

    # Dominio de la frecuencia (espectro)
    freq = rfftfreq(len(t), 1/fs)
    spectrum = np.abs(rfft(signal))

    ax2.plot(freq, spectrum, color='r')
    ax2.set_title(f'{title} - Dominio de la Frecuencia')
    ax2.set_xlabel('Frecuencia (Hz)')
    ax2.set_ylabel('Magnitud')
    ax2.set_xlim(0, 2 * fc)  # Mostrar hasta 2*fc para claridad
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

# Resultados
plot_signal(mod_rect, "DSB-SC con Pulso Rectangular")
plot_signal(mod_cos, "DSB-SC con Señal Cosenoidal")
```
