**Series de Fourier**

*  Las Series de Fourier se utilizan para representar funciones periódicas como una suma de senos y cosenos (o exponenciales complejas).

**Serie de Fourier Exponencial:**

*  Representa una función periódica $f(t)$ con período $T_0$ como: $f(t) = \sum_{n=-\infty}^{\infty} c_n e^{j n \omega_0 t}$ donde $\omega_0 = 2\pi / T_0$ y los coeficientes $c_n$ se calculan como: $c_n = \frac{1}{T_0} \int_{t_0}^{t_0+T_0} f(t) e^{-j n \omega_0 t} dt$


**Serie de Fourier Trigonométrica:**

*  Representa una función periódica $f(t)$ con período $T_0$ como: $f(t) = a_0 + \sum_{n=1}^{\infty} (a_n \cos(n \omega_0 t) + b_n \sin(n \omega_0 t))$ donde $\omega_0 = 2\pi / T_0$ y los coeficientes se calculan como: $a_0 = \frac{1}{T_0} \int_{t_0}^{t_0+T_0} f(t) dt$ $a_n = \frac{2}{T_0} \int_{t_0}^{t_0+T_0} f(t) \cos(n \omega_0 t) dt$ $b_n = \frac{2}{T_0} \int_{t_0}^{t_0+T_0} f(t) \sin(n \omega_0 t) dt$


**Serie de Fourier Compacta:**
*  Es otra forma de la serie trigonométrica, expresando la suma de senos y cosenos como una sola sinusoide con fase.


**Transformada de Fourier (Continuous-Time Fourier Transform - CTFT)**

*   La Transformada de Fourier se utiliza para representar funciones no periódicas (o aperiódicas) como una integral de exponenciales complejas. Representa la función en el dominio de la frecuencia.

**Definición:**

  *   La transformada de Fourier de una función $f(t)$ es: $F(\omega) = \int_{-\infty}^{\infty} f(t) e^{-j \omega t} dt$


**Transformada Inversa de Fourier:**

*   La transformada inversa es: $f(t) = \frac{1}{2\pi} \int_{-\infty}^{\infty} F(\omega) e^{j \omega t} d\omega$

**Transformada de Fourier en Tiempo Discreto (Discrete-Time Fourier Transform - DTFT)**

* La DTFT se utiliza para representar secuencias discretas (señales definidas solo en instantes de tiempo discretos) como una integral de exponenciales complejas en el dominio de la frecuencia. El dominio de la frecuencia es continuo y periódico.


**Definición:**
* La DTFT de una secuencia $x[n]$ es: $X(e^{j\omega}) = \sum_{n=-\infty}^{\infty} x[n] e^{-j \omega n}$

**Transformada Inversa de Fourier en Tiempo Discreto:**
* La transformada inversa es: $x[n] = \frac{1}{2\pi} \int_{-\pi}^{\pi} X(e^{j\omega}) e^{j \omega n} d\omega$

**Transformada Discreta de Fourier (Discrete Fourier Transform - DFT)**

* La DFT se utiliza para representar secuencias discretas y finitas como una suma finita de exponenciales complejas discretas. Se utiliza comúnmente en procesamiento digital de señales y se puede calcular eficientemente usando el algoritmo de la Transformada Rápida de Fourier (FFT).

**Definición:**
* La DFT de una secuencia de $N$ muestras $x[n]$ es: $X[k] = \sum_{n=0}^{N-1} x[n] e^{-j \frac{2\pi}{N} k n}$, para $k = 0, 1, \ldots, N-1$

**Transformada Inversa Discreta de Fourier:**
* La transformada inversa es: $x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j \frac{2\pi}{N} k n}$, para $n = 0, 1, \ldots, N-1$

**Semejanzas:**

* Todas representan señales en términos de sus componentes de frecuencia.
* Todas utilizan exponenciales complejas (o senos y cosenos) como base para la representación.
* Conceptualmente, buscan descomponer una señal en sus "frecuencias constituyentes".

**Diferencias:**

* **Dominio de la Señal:**
    * Las Series de Fourier trabajan con señales continuas y periódicas. La CTFT trabaja con señales continuas y no periódicas. La DTFT trabaja con señales discretas y no periódicas. La DFT trabaja con señales discretas y finitas.
* **Dominio de la Frecuencia:**
    * Para las Series de Fourier, el espectro de frecuencia es discreto. Para la CTFT, el espectro de frecuencia es continuo y no periódico. Para la DTFT, el espectro de frecuencia es continuo y periódico. Para la DFT, el espectro de frecuencia es discreto y finito.
* **Tipo de Representación:**
    * Las Series de Fourier representan la señal como una suma infinita. La CTFT y la DTFT representan la señal como una integral. La DFT representa la señal como una suma finita.
* **Aplicación:**
    * Las Series de Fourier se usan para analizar señales periódicas. La CTFT se usa para analizar señales continuas no periódicas. La DTFT se usa para analizar señales discretas (pero el espectro es continuo). La DFT se usa para analizar digitalmente señales discretas y finitas, y es la base para la implementación práctica de análisis de frecuencia en sistemas digitales.

# Combinación de Tiempo y Espectro en Señales

Cuando se combinan las características del **tiempo** (continuo o discreto) con las del **espectro** (continuo o discreto), se obtienen **cuatro combinaciones posibles**. Estas combinaciones describen cómo una señal puede comportarse tanto en el dominio del tiempo como en el dominio de la frecuencia. A continuación se detallan:

---

###  1. Señal en **tiempo continuo** con **espectro continuo**

- **Ejemplo**: Señales analógicas como una sinusoide pura o una suma de senos.
- **Transformada usada**: Transformada de Fourier continua.
- **Aplicación típica**: Audio analógico, señales de radio.
- **Características**: Energía distribuida continuamente en frecuencia.

---

###  2. Señal en **tiempo continuo** con **espectro discreto**

- **Ejemplo**: Señales periódicas en tiempo continuo (como una sinusoide de frecuencia fija).
- **Transformada usada**: Serie de Fourier.
- **Aplicación típica**: Osciladores, sistemas con señales periódicas.
- **Características**: Componentes armónicos en frecuencias discretas.

---

###  3. Señal en **tiempo discreto** con **espectro continuo**

- **Ejemplo**: Señales no periódicas en tiempo discreto (como una señal muestreada que no es periódica).
- **Transformada usada**: Transformada de Fourier discreta en tiempo (DTFT).
- **Aplicación típica**: Señales digitales capturadas de fenómenos naturales.
- **Características**: El espectro es continuo dentro del intervalo $[-\pi, \pi]$.

---

###  4. Señal en **tiempo discreto** con **espectro discreto**

- **Ejemplo**: Señales periódicas en tiempo discreto.
- **Transformada usada**: Serie de Fourier discreta (DFS) o DFT.
- **Aplicación típica**: Procesamiento digital de señales periódicas, compresión.
- **Características**: Espectro con un número finito de componentes.

# Fast Fourier Transform (FFT) y su utilidad

La **Fast Fourier Transform (FFT)** o **Transformada Rápida de Fourier** es un **algoritmo eficiente** para calcular la **Transformada Discreta de Fourier (DFT)** de una secuencia. Mientras que la DFT permite transformar una señal del dominio del tiempo al dominio de la frecuencia, su cálculo directo es **computacionalmente costoso** para señales largas.

---

##  ¿Qué es la DFT?

La **Transformada Discreta de Fourier** para una secuencia de $N$ puntos se define como:

$$[
X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j \frac{2\pi}{N}kn}, \quad k = 0, 1, ..., N-1
]$$

Esto requiere realizar $N$ sumas, cada una con $N$ multiplicaciones ⇒ **costo computacional: $\mathcal{O}(N^2)$**.

---

# ¿Qué hace la FFT?

La **FFT reduce drásticamente el número de operaciones necesarias** para obtener la DFT, pasando de $\mathcal{O}(N^2)$ a **$\mathcal{O}(N \log N)$**.

Para ello, aprovecha propiedades de simetría y periodicidad de los factores exponenciales (llamados *twiddle factors*) y divide recursivamente el problema en subproblemas más pequeños.

---

# ¿Cómo funciona la FFT? (Algoritmo de Cooley-Tukey)

El algoritmo **más común** de FFT es el **de Cooley-Tukey**, que funciona mejor cuando $N$ es una potencia de 2. La idea principal es:

1. **Dividir** la secuencia original $x[n]$ en dos subsecuencias:
   - Los elementos **pares**: $x[0], x[2], x[4], \dots$
   - Los elementos **impares**: $x[1], x[3], x[5], \dots$

2. Aplicar recursivamente la FFT a ambas subsecuencias (de longitud $N/2$).

3. Combinar los resultados usando los factores de rotación:

$$[
X[k] = E[k] + W_N^k \cdot O[k] \\
X[k + N/2] = E[k] - W_N^k \cdot O[k]
]$$

donde:
- $E[k]$ es la FFT de la parte par.
- $O[k]$ es la FFT de la parte impar.
- $W_N^k = e^{-j\frac{2\pi k}{N}}$ es el *twiddle factor*.

---

# Complejidad Computacional

| Método        | Complejidad               |
|---------------|---------------------------|
| DFT directa   | $\mathcal{O}(N^2)$        |
| FFT (Cooley-Tukey) | $\mathcal{O}(N \log N)$ |

Esto significa que para señales largas (por ejemplo, $N = 1024$, $2048$, etc.), la FFT es **muchísimo más rápida** que el cálculo directo de la DFT.

---

# Utilidad de la FFT

- **Procesamiento de señales** en tiempo real (audio, radar, comunicaciones).
- **Análisis espectral** rápido y eficiente.
- **Compresión** y **filtrado** de señales digitales.
- **Reconstrucción de señales** en aplicaciones médicas e industriales.

# Transformada de Fourier de señales específicas

Usamos la **convención de Fourier bilateral**:

$$
\boxed{
X(f) = \int_{-\infty}^{\infty} x(t) e^{-j 2\pi f t} dt
}
$$

---

**a)** $x(t) = e^{-a|t|}, \quad a \in \mathbb{R}^+$

Esta función es par. La TF se calcula como:

$$
X(f) = 2 \int_0^\infty e^{-a t} \cos(2\pi f t) dt = \frac{2a}{a^2 + (2\pi f)^2}
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{e^{-a|t|}\} = \frac{2a}{a^2 + (2\pi f)^2}
}
$$

---

**b)** $x(t) = \cos(2\pi f_c t), \quad f_c \in \mathbb{R}$

Usando la identidad de Euler:

$$
\cos(2\pi f_c t) = \frac{1}{2} \left( e^{j 2\pi f_c t} + e^{-j 2\pi f_c t} \right)
$$

$$
X(f) = \frac{1}{2}\left[\delta(f - f_c) + \delta(f + f_c)\right]
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{\cos(2\pi f_c t)\} = \frac{1}{2}\left[\delta(f - f_c) + \delta(f + f_c)\right]
}
$$

---

**c)** $x(t) = \sin(2\pi f_s t), \quad f_s \in \mathbb{R}$

$$
\sin(2\pi f_s t) = \frac{1}{2j} \left( e^{j 2\pi f_s t} - e^{-j 2\pi f_s t} \right)
$$

$$
X(f) = \frac{1}{2j}\left[\delta(f - f_s) - \delta(f + f_s)\right]
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{\sin(2\pi f_s t)\} = \frac{1}{2j}\left[\delta(f - f_s) - \delta(f + f_s)\right]
}
$$

---

**d)** Sea $x(t) = f(t)\cos(\omega_c t)$, donde $f(t) \in \mathbb{R}$ o $\mathbb{C}$, y $\omega_c \in \mathbb{R}$.

**Transformada de Fourier** (propiedad de modulación):  
$$
\mathcal{F}\{x(t)\} = \frac{1}{2} F(\omega - \omega_c) + \frac{1}{2} F(\omega + \omega_c),
$$  
donde $F(\omega) = \mathcal{F}\{f(t)\}$.

**Nota:** Si $f(t)$ no está definida explícitamente, no puede evaluarse numéricamente.

---

**e)** $x(t) = e^{-a t^2}, \quad a \in \mathbb{R}^+$

Esta es una **gaussiana**. La TF también es gaussiana:

$$
X(f) = \sqrt{\frac{\pi}{a}} \cdot e^{-\frac{(\pi f)^2}{a}}
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\{e^{-a t^2}\} = \sqrt{\frac{\pi}{a}} \cdot e^{-\frac{(\pi f)^2}{a}}
}
$$

---

**f)** $x(t) = A \cdot \text{rect}\left(\frac{t}{d}\right), \quad A, d \in \mathbb{R}$

La función $\text{rect}(t/d)$ vale 1 entre $-d/2$ y $d/2$:

$$
X(f) = A \int_{-d/2}^{d/2} e^{-j 2\pi f t} dt = A \cdot d \cdot \text{sinc}(f d)
$$

**Resultado:**

$$
\boxed{
\mathcal{F}\left\{A \cdot \text{rect}\left(\frac{t}{d}\right)\right\} = A \cdot d \cdot \text{sinc}(f d)
}
$$

donde $\text{sinc}(x) = \frac{\sin(\pi x)}{\pi x}$.

# Modulación por Amplitud con Detección Coherente (DSB-SC)

La **modulación por amplitud con portadora suprimida** (DSB-SC, *Double Sideband Suppressed Carrier*) es un esquema de modulación donde la señal mensaje modula directamente una portadora mediante multiplicación, **eliminando la componente de la portadora** en la señal transmitida. A diferencia de la AM convencional, **requiere detección coherente** en el receptor, es decir, la portadora local debe estar sincronizada en fase y frecuencia para una demodulación correcta.

---

## **Características Clave**

**Señal modulada:**

\[
s(t) = m(t) \cdot \cos(2\pi f_c t)
\]

donde \( m(t) \) es la señal mensaje y \( f_c \) es la frecuencia de la portadora.

**Demodulación:**
1. Multiplicar por \( \cos(2\pi f_c t) \) (portadora local).
2. Aplicar un filtro pasa bajos para recuperar \( m(t) \).

---

### **Ventajas**
- Mayor eficiencia espectral (no se desperdicia potencia en la portadora).
- Ideal para sistemas con restricciones de ancho de banda.

### **Desventajas**
- Requiere sincronización precisa en el receptor (complejidad aumentada).

---

## Aplicaciones
- Comunicaciones de banda lateral única (SSB).
- Radio definida por software (SDR).
- Transmisión de señales analógicas y digitales en entornos con interferencia controlada.

---

## Simulación en Python

El siguiente código genera y visualiza una señal DSB-SC, tanto en el dominio del tiempo como en el de la frecuencia, usando dos tipos de señales mensaje:

- **Pulso rectangular** (para ver el efecto en señales no periódicas).
- **Señal cosenoidal** (para analizar el espectro típico).


````python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la señal
f_carrier = 800       # Frecuencia de la portadora (Hz)
f_message = 40        # Frecuencia de la señal de mensaje (Hz)
A_msg = 1.2           # Amplitud de la señal mensaje
mod_index = 0.6       # Índice de modulación (definido por el usuario)

# Tiempo de simulación
start_t = 0
end_t = 6 / f_message   # Duración: 6 periodos del mensaje
sample_rate = 6 * f_carrier
t = np.arange(start_t, end_t, 1 / sample_rate)

# Amplitud de la portadora para cumplir con el índice de modulación
A_carrier = A_msg / mod_index

# Señal portadora
carrier = A_carrier * np.sin(2 * np.pi * f_carrier * t)

# Mensaje 1: Pulso rectangular
msg_pulse = np.zeros_like(t)
msg_pulse[int(0.25 * len(t)):int(0.55 * len(t))] = A_msg

# Mensaje 2: Coseno
msg_cosine = A_msg * np.cos(2 * np.pi * f_message * t)

# Modulación AM
am_pulse = (1 + msg_pulse / A_carrier) * carrier
am_cosine = (1 + msg_cosine / A_carrier) * carrier

# Detección coherente (multiplicación por portadora)
coh_pulse = am_pulse * carrier
coh_cosine = am_cosine * carrier

# Filtro pasa-bajo simple (ventana móvil)
def lowpass_filter(signal, N=120):
    return np.convolve(signal, np.ones(N)/N, mode='same')

recovered_pulse = lowpass_filter(coh_pulse)
recovered_cosine = lowpass_filter(coh_cosine)

# FFT en frecuencia
def plot_fft(signal, label, color):
    fft_data = np.fft.rfft(signal)
    freq = np.fft.rfftfreq(len(signal), 1 / sample_rate)
    plt.plot(freq, np.abs(fft_data), label=label, color=color)

# ---- Gráficas ----
plt.figure(figsize=(14, 12))

# Tiempo: Portadora + Pulso
plt.subplot(4, 2, 1)
plt.plot(t, carrier, label='Portadora', color='gray', alpha=0.6)
plt.plot(t, msg_pulse, label='Mensaje: Pulso', color='darkorange')
plt.title('Pulso rectangular + Portadora')
plt.xlabel('Tiempo (s)')
plt.legend()
plt.grid(True)

# Tiempo: Señal AM con pulso
plt.subplot(4, 2, 2)
plt.plot(t, am_pulse, color='darkorange')
plt.title(f'Señal AM (Pulso) - Índice m = {mod_index}')
plt.xlabel('Tiempo (s)')
plt.grid(True)

# Tiempo: Portadora + Coseno
plt.subplot(4, 2, 3)
plt.plot(t, carrier, label='Portadora', color='gray', alpha=0.6)
plt.plot(t, msg_cosine, label='Mensaje: Coseno', color='steelblue')
plt.title('Coseno + Portadora')
plt.xlabel('Tiempo (s)')
plt.legend()
plt.grid(True)

# Tiempo: Señal AM con coseno
plt.subplot(4, 2, 4)
plt.plot(t, am_cosine, color='steelblue')
plt.title(f'Señal AM (Coseno) - Índice m = {mod_index}')
plt.xlabel('Tiempo (s)')
plt.grid(True)

# Frecuencia: FFT Pulso
plt.subplot(4, 2, 5)
plot_fft(msg_pulse, 'Mensaje: Pulso', 'darkorange')
plot_fft(carrier, 'Portadora', 'gray')
plot_fft(am_pulse, 'AM: Pulso', 'firebrick')
plt.title('Espectro - Pulso')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Magnitud')
plt.legend()
plt.grid(True)

# Frecuencia: FFT Coseno
plt.subplot(4, 2, 6)
plot_fft(msg_cosine, 'Mensaje: Coseno', 'steelblue')
plot_fft(carrier, 'Portadora', 'gray')
plot_fft(am_cosine, 'AM: Coseno', 'firebrick')
plt.title('Espectro - Coseno')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Magnitud')
plt.legend()
plt.grid(True)

# Tiempo: Recuperación pulso
plt.subplot(4, 2, 7)
plt.plot(t, recovered_pulse, color='darkorange')
plt.title('Mensaje Recuperado - Pulso')
plt.xlabel('Tiempo (s)')
plt.grid(True)

# Tiempo: Recuperación coseno
plt.subplot(4, 2, 8)
plt.plot(t, recovered_cosine, color='steelblue')
plt.title('Mensaje Recuperado - Coseno')
plt.xlabel('Tiempo (s)')
plt.grid(True)

plt.tight_layout()
plt.show()
````

### Modulación de Amplitud (AM): Forma de la Señal y Espectro

#### 1. Señal Portadora y Señal Modulada en AM

La señal portadora está dada por:

$$
c(t) = A_c \cos(2\pi f_c t)
$$

Al modularla en amplitud con una señal mensaje \$m(t)\$, se obtiene la señal AM con envolvente:

$$
y(t) = \underbrace{\left(1 + \frac{m(t)}{A_c}\right)}_{\text{Envolvente}} A_c \cos(2\pi f_c t) = \underbrace{A_c \cos(2\pi f_c t)}_{\text{Portadora}} + \underbrace{m(t) \cos(2\pi f_c t)}_{\text{Señal modulada}}
$$

**Notas clave:**

* La **envolvente** \$\left(1 + \frac{m(t)}{A\_c}\right)\$ contiene la información de \$m(t)\$.
* El término \$A\_c \cos(2\pi f\_c t)\$ es la **portadora no modulada**.
* El término \$m(t) \cos(2\pi f\_c t)\$ representa la **señal mensaje trasladada en frecuencia**.

---

#### 2. Espectro de la Señal AM

Aplicando la propiedad de modulación de la transformada de Fourier:

$$
\mathcal{F} \{ m(t) \cos(2\pi f_c t) \} = \frac{1}{2} \left[ M(f - f_c) + M(f + f_c) \right]
$$

Por lo tanto, el espectro de \$y(t)\$ es:

$$
Y(f) = \underbrace{\frac{A_c}{2} \left[\delta(f - f_c) + \delta(f + f_c)\right]}_{\text{Componentes de la portadora}} + \underbrace{\frac{1}{2} \left[ M(f - f_c) + M(f + f_c) \right]}_{\text{Bandas laterales}}
$$

**Interpretación del espectro:**

* **Componentes discretas en** \$\pm f\_c\$: representan la portadora (impulsos delta).
* **Bandas laterales:**

  * Ubicadas en \$f\_c \pm f\$ y \$-f\_c \pm f\$ (simétricas).
  * Contienen el espectro de \$m(t)\$ desplazado alrededor de \$\pm f\_c\$.



## Demodulación de Señales AM (Caso θ₀ = 0)

### 1. Señal Recibida

Cuando el desfase inicial θ₀ es cero, la señal recibida r(t) corresponde a una señal AM con portadora suprimida:

$$
r(t) = A_1 \cdot m(t) \cdot \cos(2\pi f_0 t)
$$

donde:
* \$A_1\$: Amplitud de la portadora
* \$m(t)\$: Señal moduladora (información)
* \$f_0\$: Frecuencia de la portadora

---

### 2. Proceso de Mezclado

El demodulador multiplica r(t) por la portadora local cos(2πf₀t):

$$
r(t) \cdot \cos(2\pi f_0 t) = A_1 \cdot m(t) \cdot \cos^2(2\pi f_0 t)
$$

---

### 3. Aplicación de Identidad Trigonométrica

Usando la identidad:

$$
\cos^2(\theta) = \frac{1}{2} + \frac{1}{2} \cos(2\theta)
$$

obtenemos:

$$
r(t) \cdot \cos(2\pi f_0 t) = \underbrace{\frac{A_1}{2} m(t)}_{\text{Banda base}} + \underbrace{\frac{A_1}{2} m(t) \cos(4\pi f_0 t)}_{\text{Componente en 2f₀}}
$$

**Componentes:**
1. Señal original en banda base
2. Señal modulada en 2f₀

---

### 4. Filtrado Pasa-Bajos

Un LPF con fₚ < 2f₀ elimina el componente de alta frecuencia:

$$
\text{Salida LPF} = \frac{A_1}{2} m(t)
$$

**Especificaciones del filtro:**
* Ancho de banda ≥ ancho de banda de m(t)
* Atenuación ≥ 40 dB en 2f₀

---

### 5. Recuperación de la Señal

Ganancia de \$\frac{2}{A_1}\$ para obtener m(t) original:

$$
m(t) = \frac{2}{A_1} \cdot \frac{A_1}{2} m(t)
$$

---

### Diagrama de Bloques

```mermaid
Señal AM → [Mezclador] → [LPF] → [Ajuste de Ganancia] → m(t)
                      ↓
                  Portadora local (cos(2πf₀t))
```






````python
#graficas de etapas principales
import numpy as np
import matplotlib.pyplot as plt
import soundfile as sf
import scipy.signal as signal
from scipy.fft import fft, fftfreq, ifft

import IPython.display as ipd

# 1. Cargar fragmento de audio original (mensaje m(t))
mensaje, fs = sf.read('fragmento.wav')
t = np.linspace(0, len(mensaje)/fs, len(mensaje), endpoint=False)

# 2. Generar portadora
fc = 10000  # frecuencia de portadora en Hz
portadora = np.cos(2 * np.pi * fc * t)

# 3. Modulación AM: y(t) = (1 + m(t)) * c(t)
modulada = (1 + mensaje) * portadora

# 4. Demodulación coherente: multiplicar por cos(2πf₀t)
mezcla = modulada * portadora  # señal multiplicada

# 5. Filtrado ideal espectral (pasa bajas)
def filtro_ideal_espectral(signal, fs, cutoff):
    N = len(signal)
    fft_signal = fft(signal)
    freqs = fftfreq(N, 1/fs)

    # Crear máscara de paso bajas
    filtro = np.abs(freqs) < cutoff
    fft_filtrada = fft_signal * filtro

    return np.real(ifft(fft_filtrada)), freqs, fft_signal, fft_filtrada

recuperada, freqs, mezcla_fft, filtrada_fft = filtro_ideal_espectral(mezcla, fs, cutoff=8000)

# --- Graficar todas las etapas ---
def graficar_senal_t_f(senal, fs, titulo, tiempo, freqs=None):
    N = len(senal)
    if freqs is None:
        freqs = fftfreq(N, 1/fs)
    espectro = np.abs(fft(senal))

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(tiempo, senal)
    plt.title(f"{titulo} - Tiempo")
    plt.xlabel("t [s]")
    plt.ylabel("Amplitud")

    plt.subplot(1, 2, 2)
    plt.plot(freqs[:N//2], espectro[:N//2])
    plt.title(f"{titulo} - Espectro (Magnitud)")
    plt.xlabel("Frecuencia [Hz]")
    plt.ylabel("|X(f)|")
    plt.tight_layout()
    plt.show()

# Graficar y reproducir cada etapa
graficar_senal_t_f(mensaje, fs, "Mensaje original", t)
display(ipd.Audio(mensaje, rate=fs))

graficar_senal_t_f(portadora, fs, "Portadora", t)
display(ipd.Audio(portadora, rate=fs))

graficar_senal_t_f(modulada, fs, "Señal Modulada AM", t)
display(ipd.Audio(modulada, rate=fs))

graficar_senal_t_f(mezcla, fs, "Producto con portadora (Mezcla)", t)
display(ipd.Audio(mezcla, rate=fs))

graficar_senal_t_f(recuperada, fs, "Señal Recuperada (Filtrado ideal)", t, freqs)
display(ipd.Audio(recuperada, rate=fs))
````

## 1. Análisis de Distorsión Armónica (THD)  

**Concepto Clave**  
La Distorsión Armónica Total (THD) evalúa cuánto se desvía una señal periódica de su forma sinusoidal ideal, cuantificando la contribución relativa de sus componentes armónicos respecto a la frecuencia fundamental.  

**Ecuación Fundamental**  
$$  
\text{THD} (\%) = 100 \times \frac{\sqrt{\sum_{n=2}^{\infty} V_n^2}}{V_1}  
$$  
- **Variables**:  
  - \( V_1 \): Amplitud de la frecuencia fundamental (60 Hz en sistemas eléctricos típicos).  
  - \( V_n \): Amplitud del n-ésimo armónico (120 Hz, 180 Hz, etc.).  

**Procedimiento de Cálculo**  
1. **Descomposición Espectral**: Utilizar la FFT para obtener el espectro de frecuencias de la señal.  
2. **Identificación de Armónicos**: Localizar los picos en múltiplos enteros de la frecuencia fundamental.  
3. **Cálculo de Contribuciones**: Sumar las energías de los armónicos y normalizar respecto a la fundamental.  

---  

## 2. Factor de Potencia y su Relación con Armónicos  

**Definición Operacional**  
El Factor de Potencia (FP) determina la eficiencia energética de un sistema, relacionando la potencia útil (activa) con la potencia total suministrada (aparente). En presencia de armónicos, se expresa como:  
$$  
\text{FP} = \frac{P_{\text{activa}}}{\sqrt{P_{\text{activa}}^2 + Q^2 + D^2}}  
$$  
- **Componentes**:  
  - \( Q \): Potencia reactiva (asociada a desfases).  
  - \( D \): Potencia distorsionante (debida a armónicos).  

**Relación THD-FP**  
Para corrientes distorsionadas:  
$$  
\text{FP} \approx \frac{\cos \phi}{\sqrt{1 + \text{THD}_I^2}}  
$$  
Donde $( \text{THD}_I )$ es el THD de la corriente y $( \cos \phi )$ el factor de desplazamiento.  

---  
  



````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq
from scipy.signal import lfilter

# Configuración de estilo para gráficos
plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['axes.grid'] = True

def configurar_ejes(ax, title, xlabel, ylabel):
    """Configuración estandarizada de ejes para gráficos"""
    ax.set_title(title, fontsize=14, pad=20)
    ax.set_xlabel(xlabel, fontsize=12)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.tick_params(axis='both', which='major', labelsize=10)

def calcular_thd(signal, fs, n_armonicos=15):
    """
    Calcula el THD de una señal considerando los primeros n_armonicos

    Args:
        signal: Señal temporal de entrada
        fs: Frecuencia de muestreo
        n_armonicos: Número de armónicos a considerar

    Returns:
        thd: Valor de THD en porcentaje
        freqs: Vector de frecuencias
        fft_vals: Magnitudes de la FFT
    """
    N = len(signal)

    # Cálculo de la FFT
    fft_vals = 2*np.abs(fft(signal))/N  # Normalización y corrección de amplitud
    freqs = fftfreq(N, 1/fs)

    # Solo frecuencias positivas
    mask = (freqs > 0) & (freqs <= (n_armonicos+1)*60)
    freqs = freqs[mask]
    fft_vals = fft_vals[mask]

    # Identificación de componentes
    idx_fundamental = np.argmin(np.abs(freqs - 60))  # 60 Hz fundamental
    fundamental = fft_vals[idx_fundamental]

    # Excluimos la fundamental y DC
    mask_armonicos = ~np.isclose(freqs, 60, atol=1)
    armonicos = fft_vals[mask_armonicos]

    thd = 100 * np.sqrt(np.sum(armonicos**2)) / fundamental
    return thd, freqs, fft_vals

def graficar_fft(freqs, fft_vals, ax, color='b'):
    """Grafica el espectro de frecuencia"""
    ax.stem(freqs, fft_vals, linefmt=f'{color}-', markerfmt=f'{color}o', basefmt=' ')
    ax.set_xlim(0, 1000)
    ax.set_ylim(0, 1.1*np.max(fft_vals))

# Parámetros de simulación
f_linea = 60  # Frecuencia de red (Hz)
w = 2 * np.pi * f_linea
T = 1/f_linea
muestras_por_ciclo = 1000
t = np.linspace(0, 2*T, 2*muestras_por_ciclo)
fs = 1/(t[1]-t[0])  # Frecuencia de muestreo

# Generación de señal rectificada
Vin = np.abs(np.sin(w * t))  # Rectificador de onda completa ideal

# Caso 1: Carga puramente resistiva
R = 10  # Ohms
I_R = Vin / R

# Caso 2: Carga RC en serie
R = 10  # Ohms
C = 100e-6  # Faradios
tau = R * C

# Implementación de filtro pasa-bajos (modelo de capacitor)
alpha = (t[1]-t[0]) / (tau + (t[1]-t[0]))
I_RC = np.zeros_like(Vin)
for i in range(1, len(Vin)):
    I_RC[i] = I_RC[i-1] + alpha * (Vin[i]/R - I_RC[i-1])

# Cálculo de THD
THD_R, freqs_R, fft_R = calcular_thd(I_R, fs)
THD_RC, freqs_RC, fft_RC = calcular_thd(I_RC, fs)

# Visualización
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# Señales en el tiempo
ax1.plot(t, I_R, label=f'Carga R (THD = {THD_R:.2f}%)')
ax1.plot(t, I_RC, 'r', label=f'Carga RC (THD = {THD_RC:.2f}%)')
configurar_ejes(ax1, 'Corriente en el Rectificador', 'Tiempo [s]', 'Corriente [A]')
ax1.legend()

# Espectros de frecuencia
ax2.stem(freqs_R, fft_R, linefmt='b-', markerfmt='bo', label='Carga R')
ax2.stem(freqs_RC, fft_RC, linefmt='r-', markerfmt='ro', label='Carga RC')
configurar_ejes(ax2, 'Espectro de Frecuencia', 'Frecuencia [Hz]', 'Amplitud [A]')
ax2.set_xlim(0, 1000)
ax2.legend()

plt.tight_layout()
plt.show()

# Resultados numéricos
print("\nRESULTADOS:")
print(f"THD carga resistiva: {THD_R:.2f}%")
print(f"THD carga RC: {THD_RC:.2f}%")
print(f"Relación THD RC/R: {THD_RC/THD_R:.2f}")
````

    1. Reducción del THD (menor potencia distorsionante)
    2. Pequeño desplazamiento por la componente reactiva

### 4. Análisis Comparativo
| Parámetro       | Carga Resistiva | Carga RC     | Mejora |
|----------------|----------------|-------------|-------|
| THD típico     | 30-45%         | 5-15%       | 70-85%|
| Factor Potencia| 0.90-0.95      | 0.97-0.99   | 3-8%  |
| Armónicos      | Hasta 15° orden| Hasta 5° orden| -    |

### 5. Implicaciones Prácticas
1. **Diseño de filtros**:
   - La carga RC actúa como filtro pasivo natural
   - Puede complementarse con inductancias para mayor suavizado

2. **Eficiencia energética**:
   - La reducción de THD minimiza pérdidas por calentamiento
   - Mejor FP reduce la corriente aparente requerida

3. **Compatibilidad electromagnética**:
   - Menor emisión de armónicos a la red eléctrica
   - Cumplimiento más fácil con normativas IEEE 519

### 6. Limitaciones y Consideraciones
- **Compromiso diseño RC**:
  - Valores altos de C aumentan corriente transitoria inicial
  - Requiere balance entre suavizado y tamaño/peso del capacitor
- **Análisis completo**:
  - En aplicaciones reales considerar:
    - Resistencia parásita de cables
    - Modelado preciso de diodos
    - Efectos térmicos

Esta discusión técnica evidencia cómo la selección adecuada de componentes pasivos puede optimizar simultáneamente múltiples parámetros de calidad de potencia en sistemas electrónicos de potencia.

````python
%%writefile detector.py
import streamlit as st
from pytube import YouTube
import os
import librosa
import numpy as np

# Modelo de ejemplo
class DummyModel:
    def predict(self, audio):
        return "Masculino" if np.mean(audio) > 0 else "Femenino"

model = DummyModel()

def app():
    st.title("Detector de Género en YouTube")
    url = st.text_input("Ingresa un enlace de YouTube")

    if st.button("Detectar Género") and url:
        try:
            yt = YouTube(url)
            stream = yt.streams.filter(only_audio=True).first()
            audio_file = stream.download(filename="audio.mp4")

            audio, sr = librosa.load(audio_file, sr=16000, mono=True, duration=10)
            prediction = model.predict(audio)

            st.audio(audio_file)
            st.write(f"**Género Detectado:** {prediction}")
            st.markdown(f"[Ver video en YouTube]({url})")

            os.remove(audio_file)
        except Exception as e:
            st.error(f"Error al procesar el enlace: {e}")

    st.markdown("[Ver notebook original](https://github.com/Ebl14/senales_y_sistemas/blob/main/youtube_detector.ipynb)")
````
